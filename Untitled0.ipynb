```python

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def encode_classical_data(x, y):
    # Normalize classical input to [0, 1]
    x = np.clip(x, 0, 1)
    y = np.clip(y, 0, 1)

    # Convert data into rotation angles
    theta = np.pi * x
    phi = 2 * np.pi * y

    # Build quantum circuit
    qc = QuantumCircuit(1)

    # Data encoding
    qc.ry(theta, 0)
    qc.rz(phi, 0)

    # Calculate ideal statevector
    state = Statevector.from_instruction(qc)

    return qc, state

# Example
x = 0.5
y = 0.25

circuit, state = encode_classical_data(x, y)

print(circuit.draw())
print("Encoded state:", state.data)
```

```python

# ============================================================
# BLOCH SPHERE & DATA ENCODING VISUALIZER
# Interactive Quantum Computing Web App
# Google Colab + Gradio + Plotly
# ============================================================

!pip -q install --upgrade gradio plotly numpy

import numpy as np
import plotly.graph_objects as go
import gradio as gr
import math

# ------------------------------------------------------------
# 1. BLOCH SPHERE VISUALIZATION
# ------------------------------------------------------------

def create_bloch_sphere(x, y):

    # Normalize inputs
    x = float(np.clip(x, 0, 1))
    y = float(np.clip(y, 0, 1))

    # Classical data -> quantum rotation angles
    theta = np.pi * x
    phi = 2 * np.pi * y

    # Bloch sphere coordinates
    bx = np.sin(theta) * np.cos(phi)
    by = np.sin(theta) * np.sin(phi)
    bz = np.cos(theta)

    fig = go.Figure()

    # Sphere surface
    u = np.linspace(0, 2*np.pi, 90)
    v = np.linspace(0, np.pi, 65)

    sx = np.outer(np.cos(u), np.sin(v))
    sy = np.outer(np.sin(u), np.sin(v))
    sz = np.outer(np.ones_like(u), np.cos(v))

    fig.add_trace(go.Surface(
        x=sx, y=sy, z=sz,
        colorscale=[
            [0, "#111d3c"],
            [0.5, "#174b70"],
            [1, "#17335c"]
        ],
        opacity=0.23,
        showscale=False,
        hoverinfo="skip",
        name="Bloch sphere"
    ))

    # Equator
    t = np.linspace(0, 2*np.pi, 300)
    fig.add_trace(go.Scatter3d(
        x=np.cos(t), y=np.sin(t), z=np.zeros_like(t),
        mode="lines",
        line=dict(color="#64748b", width=3, dash="dash"),
        name="Equator"
    ))

    # Latitude circles
    for lat in [-60, -30, 30, 60]:
        a = np.radians(lat)
        r = np.cos(a)
        z = np.sin(a)
        fig.add_trace(go.Scatter3d(
            x=r*np.cos(t), y=r*np.sin(t),
            z=np.full_like(t, z),
            mode="lines",
            line=dict(color="#52718f", width=1, dash="dot"),
            opacity=0.45,
            showlegend=False,
            hoverinfo="skip"
        ))

    # Longitude circles
    for angle in np.arange(0, 360, 30):
        a = np.radians(angle)
        q = np.linspace(0, np.pi, 100)
        fig.add_trace(go.Scatter3d(
            x=np.sin(q)*np.cos(a),
            y=np.sin(q)*np.sin(a),
            z=np.cos(q),
            mode="lines",
            line=dict(color="#52718f", width=1),
            opacity=0.35,
            showlegend=False,
            hoverinfo="skip"
        ))

    # Coordinate axes
    axis_colors = ["#ef476f", "#06d6a0", "#60a5fa"]
    axis_names = ["X", "Y", "Z"]

    for i, (vec, color, name) in enumerate(zip(
        [(1,0,0), (0,1,0), (0,0,1)],
        axis_colors, axis_names
    )):
        fig.add_trace(go.Scatter3d(
            x=[-1.25*vec[0], 1.25*vec[0]],
            y=[-1.25*vec[1], 1.25*vec[1]],
            z=[-1.25*vec[2], 1.25*vec[2]],
            mode="lines+text",
            line=dict(color=color, width=4),
            text=["", name],
            textposition="top center",
            showlegend=False,
            hoverinfo="skip"
        ))

    # RY rotation trajectory: north pole to intermediate state
    q = np.linspace(0, theta, 100)
    fig.add_trace(go.Scatter3d(
        x=np.sin(q),
        y=np.zeros_like(q),
        z=np.cos(q),
        mode="lines",
        line=dict(color="#fbbf24", width=5, dash="dash"),
        name="RY path"
    ))

    # RZ phase rotation trajectory
    q = np.linspace(0, phi, 120)
    fig.add_trace(go.Scatter3d(
        x=np.sin(theta)*np.cos(q),
        y=np.sin(theta)*np.sin(q),
        z=np.full_like(q, np.cos(theta)),
        mode="lines",
        line=dict(color="#c084fc", width=5),
        name="RZ path"
    ))

    # Final state vector
    fig.add_trace(go.Scatter3d(
        x=[0, bx], y=[0, by], z=[0, bz],
        mode="lines",
        line=dict(color="#22f5d0", width=9),
        name="Quantum state vector"
    ))

    # Final state marker
    fig.add_trace(go.Scatter3d(
        x=[bx], y=[by], z=[bz],
        mode="markers+text",
        marker=dict(
            size=9,
            color="#22f5d0",
            line=dict(color="white", width=2)
        ),
        text=["|ψ⟩"],
        textposition="top center",
        textfont=dict(size=17, color="white"),
        name="Final qubit state"
    ))

    # Initial state
    fig.add_trace(go.Scatter3d(
        x=[0], y=[0], z=[1],
        mode="markers+text",
        marker=dict(size=6, color="#fbbf24"),
        text=["|0⟩"],
        textposition="top center",
        textfont=dict(color="#fbbf24"),
        name="Initial state"
    ))

    fig.update_layout(
        title=dict(
            text="3D BLOCH SPHERE | QUANTUM STATE",
            font=dict(color="white", size=18),
            x=0.5
        ),
        paper_bgcolor="#080e1b",
        plot_bgcolor="#080e1b",
        font=dict(color="#dbeafe"),
        scene=dict(
            bgcolor="#080e1b",
            xaxis=dict(
                title="X",
                range=[-1.35,1.35],
                backgroundcolor="#080e1b",
                gridcolor="#243650",
                zerolinecolor="#475569"
            ),
            yaxis=dict(
                title="Y",
                range=[-1.35,1.35],
                backgroundcolor="#080e1b",
                gridcolor="#243650",
                zerolinecolor="#475569"
            ),
            zaxis=dict(
                title="Z",
                range=[-1.35,1.35],
                backgroundcolor="#080e1b",
                gridcolor="#243650",
                zerolinecolor="#475569"
            ),
            aspectmode="cube",
            camera=dict(eye=dict(x=1.65,y=1.65,z=1.3))
        ),
        height=600,
        margin=dict(l=0,r=0,t=55,b=0),
        legend=dict(
            bgcolor="rgba(10,20,40,0.7)",
            bordercolor="#243650",
            borderwidth=1
        )
    )

    return fig, theta, phi, bx, by, bz


# ------------------------------------------------------------
# 2. QUANTUM STATE AND ANALYTICS
# ------------------------------------------------------------

def visualize(x, y):

    fig, theta, phi, bx, by, bz = create_bloch_sphere(x, y)

    # Statevector after RY followed by RZ.
    # Display state with the physically irrelevant global
    # phase removed.
    alpha = np.cos(theta/2)
    beta = np.sin(theta/2) * np.exp(1j*phi)

    p0 = abs(alpha)**2
    p1 = abs(beta)**2

    state_html = f"""
    <div class="panel">
      <h3>QUANTUM STATEVECTOR</h3>
      <div class="equation">
        |ψ⟩ = {alpha:.4f}|0⟩ +
        ({beta.real:.4f} + {beta.imag:.4f}i)|1⟩
      </div>
      <p class="muted">Complex amplitude representation</p>
      <div class="grid2">
        <div class="metric">
          <small>Amplitude α</small>
          <strong>{alpha:.5f}</strong>
        </div>
        <div class="metric">
          <small>Amplitude β</small>
          <strong>{beta.real:.4f} + {beta.imag:.4f}i</strong>
        </div>
      </div>
    </div>
    """

    probability_html = f"""
    <div class="panel">
      <h3>MEASUREMENT PROBABILITIES</h3>
      <p>|0⟩ : {p0*100:.2f}%</p>
      <div class="progress">
        <div style="width:{p0*100}%;background:#22d3a6"></div>
      </div>
      <p>|1⟩ : {p1*100:.2f}%</p>
      <div class="progress">
        <div style="width:{p1*100}%;background:#818cf8"></div>
      </div>
      <div class="grid2">
        <div class="metric">
          <small>RY angle θ</small>
          <strong>{theta:.4f} rad</strong>
        </div>
        <div class="metric">
          <small>RZ angle φ</small>
          <strong>{phi:.4f} rad</strong>
        </div>
      </div>
    </div>
    """

    bloch_html = f"""
    <div class="panel">
      <h3>BLOCH VECTOR COORDINATES</h3>
      <div class="grid3">
        <div class="metric">
          <small>X</small>
          <strong>{bx:.4f}</strong>
        </div>
        <div class="metric">
          <small>Y</small>
          <strong>{by:.4f}</strong>
        </div>
        <div class="metric">
          <small>Z</small>
          <strong>{bz:.4f}</strong>
        </div>
      </div>
      <p class="muted">x² + y² + z² = {bx**2+by**2+bz**2:.5f}</p>
    </div>
    """

    circuit_html = f"""
    <div class="panel">
      <h3>QUANTUM CIRCUIT</h3>
      <div class="circuit">
        <span class="ket">|0⟩</span>
        <span class="wire"></span>
        <span class="gate ry">RY<br><small>{theta:.3f}</small></span>
        <span class="wire"></span>
        <span class="gate rz">RZ<br><small>{phi:.3f}</small></span>
        <span class="wire"></span>
        <span class="ket">|ψ⟩</span>
      </div>
      <p class="muted">
        Initial state → Y rotation → Z phase rotation → Output state
      </p>
    </div>
    """

    return fig, state_html, probability_html, bloch_html, circuit_html


# ------------------------------------------------------------
# 3. APP DESIGN
# ------------------------------------------------------------

CSS = """
body, .gradio-container {
    background: #080e1b !important;
    color: #e2e8f0 !important;
}
.gradio-container {
    max-width: 1500px !important;
}
h1 {
    color: #5eead4 !important;
    letter-spacing: 1px;
}
h2, h3 {
    color: #67e8f9 !important;
}
.panel {
    background: #111c30;
    border: 1px solid #263b59;
    border-radius: 15px;
    padding: 18px;
    margin: 8px 0;
    color: #e2e8f0;
}
.equation {
    background: #091425;
    padding: 18px;
    border-radius: 10px;
    font-size: 20px;
    color: #5eead4;
    text-align: center;
    overflow-wrap: anywhere;
}
.muted {
    color: #94a3b8;
    font-size: 12px;
}
.grid2, .grid3 {
    display: grid;
    grid-template-columns: repeat(2,1fr);
    gap: 10px;
    margin-top: 12px;
}
.grid3 {
    grid-template-columns: repeat(3,1fr);
}
.metric {
    background: #0a1629;
    border: 1px solid #263b59;
    border-radius: 10px;
    padding: 12px;
    overflow-wrap: anywhere;
}
.metric small {
    display: block;
    color: #94a3b8;
    margin-bottom: 8px;
}
.metric strong {
    color: #5eead4;
    font-size: 17px;
}
.progress {
    height: 12px;
    background: #263b59;
    border-radius: 20px;
    overflow: hidden;
}
.progress div {
    height: 100%;
    border-radius: 20px;
}
.circuit {
    display: flex;
    align-items: center;
    justify-content: center;
    gap: 0;
    padding: 22px 4px;
    overflow-x: auto;
}
.ket {
    color: #5eead4;
    font-size: 21px;
    white-space: nowrap;
}
.wire {
    min-width: 22px;
    height: 3px;
    background: #64748b;
}
.gate {
    min-width: 64px;
    padding: 12px 7px;
    border-radius: 9px;
    text-align: center;
    font-weight: bold;
    color: white;
    white-space: nowrap;
}
.ry {
    background: #2563eb;
}
.rz {
    background: #7c3aed;
}
.gate small {
    font-weight: normal;
}
"""

with gr.Blocks(
    theme=gr.themes.Base(
        primary_hue="teal",
        secondary_hue="indigo",
        neutral_hue="slate"
    ),
    css=CSS,
    title="Bloch Sphere Quantum Visualizer"
) as demo:

    gr.Markdown("""
    # ⚛ BLOCH SPHERE & DATA ENCODING VISUALIZER

    ### Classical Numbers → Quantum Rotation Gates → Qubit States

    Explore quantum superposition, phase encoding and
    quantum state visualization using interactive 3D graphics.
    """)

    with gr.Row():

        # LEFT: Input panel
        with gr.Column(scale=1, min_width=280):

            gr.Markdown("## DATA ENCODING CONTROLS")

            x_slider = gr.Slider(
                minimum=0,
                maximum=1,
                value=0.5,
                step=0.01,
                label="Input X — Superposition (RY)"
            )

            y_slider = gr.Slider(
                minimum=0,
                maximum=1,
                value=0.25,
                step=0.01,
                label="Input Y — Phase (RZ)"
            )

            gr.Markdown("### QUICK EXPERIMENTS")

            with gr.Row():
                initial_btn = gr.Button("Initial |0⟩")
                equal_btn = gr.Button("Superposition")

            with gr.Row():
                phase_btn = gr.Button("Phase Test")
                opposite_btn = gr.Button("Opposite Pole")

            gr.Markdown("""
            ### ENCODING FORMULAS

            θ = π × X

            φ = 2π × Y

            RY(θ) → RZ(φ)

            Inputs are normalized between 0 and 1.
            """)

    gr.Markdown("---")

    gr.Markdown("## 3D QUANTUM STATE VISUALIZATION")

    sphere = gr.Plot(
        label="Interactive Bloch Sphere",
        show_label=False
    )

    gr.Markdown("""
    **Legend:** Cyan = final state vector · Yellow = RY
    trajectory · Purple = RZ phase trajectory.

    Drag to rotate, scroll to zoom and use the toolbar
    to reset the 3D camera.
    """)

    gr.Markdown("---")
    gr.Markdown("## QUANTUM CIRCUIT & ANALYTICS")

    with gr.Row():
        with gr.Column():
            circuit = gr.HTML()
            state = gr.HTML()

        with gr.Column():
            probability = gr.HTML()
            bloch = gr.HTML()

    gr.Markdown("""
    ---
    ## HOW THE SYSTEM WORKS

    1. Classical data is supplied as normalized features X and Y.
    2. RY rotates the initial qubit and creates superposition.
    3. RZ applies a relative phase to the state.
    4. The resulting state is displayed on the Bloch sphere.
    5. Measurement probabilities are calculated from the amplitudes.

    **Note:** This is an ideal single-qubit state-vector
    simulation. It visualizes quantum encoding; it does not
    execute on quantum hardware.
    """)

    outputs = [sphere, state, probability, bloch, circuit]
    inputs = [x_slider, y_slider]

    # Update all visualizations when sliders change
    x_slider.change(
        fn=visualize,
        inputs=inputs,
        outputs=outputs
    )

    y_slider.change(
        fn=visualize,
        inputs=inputs,
        outputs=outputs
    )

    # Presets
    initial_btn.click(
        lambda: (0, 0),
        outputs=[x_slider, y_slider]
    )

    equal_btn.click(
        lambda: (0.5, 0),
        outputs=[x_slider, y_slider]
    )

    phase_btn.click(
        lambda: (0.5, 0.25),
        outputs=[x_slider, y_slider]
    )

    opposite_btn.click(
        lambda: (1, 0.5),
        outputs=[x_slider, y_slider]
    )

    # Initial rendering
    demo.load(
        fn=visualize,
        inputs=inputs,
        outputs=outputs
    )

# ------------------------------------------------------------
# 4. LAUNCH APPLICATION
# ------------------------------------------------------------

demo.launch(
    share=True,
    debug=True
)
```
